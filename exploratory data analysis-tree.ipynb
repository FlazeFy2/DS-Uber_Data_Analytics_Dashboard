```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify

# Exploratory Data Analysis (EDA) - Tree Chart 15 Location With Most Order
data = pd.read_csv("ncr_ride_bookings.csv")
location_col = ["Drop Location", "Pickup Location"]

for dt in location_col:
    # Count rows
    total_context = data.groupby(dt).size().reset_index(name="Count")

    # Sort and take top 15
    total_context = total_context.sort_values(by="Count", ascending=False).head(15)

    # Escape
    total_context[dt] = (total_context[dt].astype(str).str.replace("$", r"\$", regex=False))

    # Extract
    sizes = total_context["Count"]
    labels = total_context.apply(lambda x: f"{x[dt]}\n({x['Count']})", axis=1)
    # Plot Tree Map
    plt.figure(figsize=(12, 8))
    squarify.plot(sizes=sizes, label=labels, alpha=0.8)
    plt.title(f"Top 15 {dt} With Most Orders")
    plt.axis("off")
    plt.show()
```
